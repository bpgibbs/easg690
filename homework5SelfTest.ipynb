A contour line plot uses x, y, and z values to display "contour" lines on a 2-D plane to assist a person in understanding a variable change such as temperature or elevation. The axes method contour can be used to create such a figure. A filled contour will use color or shading to fill in space in relation to contours to assist with the visualization of a variable change in 2-D. The axes method contourf can be used to create a filled contour plot.

To create a basic image animation, matplotlib.pyplot and matplotlib.image must be imported. Then, the Figure and Axes objects should be created. plt.ion() should be called to turn on matplotlib interactive mode, and images are assigned from data using the image method .imread() and stored as a list. The images are then looped through with a for loop and rendered with the .draw() pyplot method, with a pause assigned for each iteration with plt.pause() for the brain to process the image. The method ax.imshow() is used for the first iteration of the loop, and subsequent iterations update the image with .set_data().

Creating a line animation is a very similar process with light differences. The number of data points to animate should be defined, and this will be used to define a range of x and y axes. Of course, x and y axes should be labled, and this can be done with plt.xlabel() and plt.ylabel(). The animation is then preformed with a for loop that plots each point as in iterates. ax.plot() is used to plot the first iteration, and the data is updated in later iterations with .set_xdata() and .set_ydata().
